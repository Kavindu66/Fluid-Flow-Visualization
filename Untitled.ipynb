```python
import numpy as np
import matplotlib.pyplot as plt
```

# PARAMETERS

```python
K = 100

U = 1 + K/200
a = 1 + K/100
m = 2 + K/50
```

# GRID

```python
x = np.linspace(-5, 5, 300)
y = np.linspace(-5, 5, 300)

X, Y = np.meshgrid(x, y)
```

# SOURCE & SINK

```python
x_source = -a
y_source = 0

x_sink = a
y_sink = 0
```

# DISTANCES

```python
r1 = np.sqrt((X - x_source)**2 + (Y - y_source)**2)
r2 = np.sqrt((X - x_sink)**2 + (Y - y_sink)**2)
```

# ANGLES

```python
theta1 = np.arctan2(Y - y_source,
                    X - x_source)

theta2 = np.arctan2(Y - y_sink,
                    X - x_sink)
```

# POTENTIAL FUNCTION (φ)

```python
phi = U*X + (m/(2*np.pi)) * (np.log(r1) - np.log(r2))
```

# STREAM FUNCTION (ψ)

```python
psi = U*Y + (m/(2*np.pi)) * (theta1 - theta2)
```

# STAGNATION POINTS

```python
x_stag = np.sqrt(a**2 + m/(np.pi*U))

stag_x = [-x_stag, x_stag]
stag_y = [0, 0]

print("Stagnation Points:")
print(f"(-{x_stag:.3f}, 0)")
print(f"({x_stag:.3f}, 0)")
```

# PLOTTING

```python
plt.figure(figsize=(10, 8))

# Streamlines
plt.contour(X, Y, psi,
            levels=50,
            colors='blue')

# Equipotential lines
plt.contour(X, Y, phi,
            levels=50,
            linestyles='dashed',
            colors='red')

# Body contour (ψ = 0)
plt.contour(X, Y, psi,
            levels=[0],
            colors='black',
            linewidths=3)

# Stagnation points
plt.scatter(stag_x, stag_y,
            color='orange',
            s=120,
            label='Stagnation Points',
            zorder=5)

# Source
plt.scatter(x_source, y_source,
            color='green',
            s=180,
            edgecolors='black',
            label='Source (+m)',
            zorder=6)

# Sink
plt.scatter(x_sink, y_sink,
            color='purple',
            s=180,
            edgecolors='black',
            label='Sink (-m)',
            zorder=6)

# Title & labels
plt.title("Rankine Oval Flow Field",
          fontsize=16)

plt.xlabel("x-axis",
           fontsize=13)

plt.ylabel("y-axis",
           fontsize=13)

plt.legend(fontsize=12)

plt.grid(True)

plt.axis('equal')

# Save graph image
plt.savefig("Rankine_Oval_Output.png",
            dpi=300)

plt.show()
```
